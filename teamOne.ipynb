## INTRODUCTION

HR analytics is the process of collecting and analyzing human resource data to enhance an organization's workforce performance. This process is also called talent analytics, people analytics, or workforce analytics.

This data analysis method takes information that HR departments regularly collect and connects it to HR and organizational goals. This approach provides clear evidence of how HR programs contribute to the organization's achievement of its objectives and strategies.

## Need for HR Analytics

Most organizations already have data that is routinely collected, so why is there a need for specialized analytics? Can HR not simply look at the data they already have? Unfortunately, raw data on its own cannot provide useful insight. It would be like looking at a large spreadsheet full of numbers and words. Once organized, compared, and analyzed, this raw data becomes valuable and can help answer important questions.

Employees are the heart of any company, and how well a company performs depends largely on its people. When employees leave, it creates several significant problems for the business:

**Key Problems Caused by Employee Turnover:**
- **High costs:** Training new hires requires substantial money and valuable time
- **Loss of expertise:** The company loses experienced employees who understand the job well
- **Reduced productivity:** Overall work output can drop significantly
- **Financial impact:** Company profits may be negatively affected

HR analytics helps organizations identify patterns in employee data to address these problems before they become costly issues.

### Description about the data

- Age: A period of employee life, measured by years from birth.

- Attrition: The departure of employees from the organization - our target variable for prediction.

- BusinessTravel: Indicates whether the employee travels on business trips, with categories including Non-Travel, Travel_Rarely, and Travel_Frequently.

- DailyRate: Employee salary for the period divided by the number of calendar days in the period.

- Department: Specifies which department the employee is working in - Research & Development, Sales, or Human Resources.

- DistanceFromHome: Measures how far the employee lives from the office location.

- Education: Educational level where 1 means 'Below College', 2 means 'College', 3 means 'Bachelor', 4 means 'Master', and 5 means 'Doctor'.

- EducationField: The field in which the employee completed their education, including Life Sciences, Medical, Marketing, Technical Degree, Human Resources, and Other.

- EmployeeCount: The number of employees working in a department (constant value of 1 for all records).

- EmployeeNumber: A unique number assigned to each current and former employee in the organization's database system.

- EnvironmentSatisfaction: Employee satisfaction with their work environment where 1 means 'Low', 2 means 'Medium', 3 means 'High', and 4 means 'Very High'.

- Gender: The gender of the employee (Male/Female).

- HourlyRate: The hourly wage rate of the employee.

- JobInvolvement: The degree to which an employee identifies with their work and actively participates in it, where 1 means 'Low', 2 means 'Medium', 3 means 'High', and 4 means 'Very High'.

- JobLevel: Job classifications that set the responsibility level and expectations of roles, often associated with pay bands and defined by impact, seniority, knowledge, and skills.

- JobRole: The specific job position of an employee, such as Sales Executive, Research Scientist, or Laboratory Technician.

- JobSatisfaction: Employee job satisfaction rate where 1 means 'Low', 2 means 'Medium', 3 means 'High', and 4 means 'Very High'.

- MaritalStatus: The marital status of the employee (Single, Married, or Divorced).

- MonthlyIncome: The total monetary value paid by the organization to an employee on a monthly basis.

- MonthlyRate: The monthly wage rate of the employee.

- NumCompaniesWorked: The number of organizations the employee worked for before joining the current organization.

- Over18: Indicates whether the employee's age is over 18 years (constant value of 'Y' for all records).

- OverTime: Indicates if an employee works more than standard hours in any day or week (Yes/No).

- PercentSalaryHike: The percentage increase in salary from the previous year.

- PerformanceRating: Annual performance evaluation where 1 means 'Low', 2 means 'Good', 3 means 'Excellent', and 4 means 'Outstanding'.

- RelationshipSatisfaction: Employee satisfaction with workplace relationships where 1 means 'Low', 2 means 'Medium', 3 means 'High', and 4 means 'Very High'.

- StandardHours: The standard number of working hours per period (constant value of 80 for all records).

- StockOptionLevel: Employee stock options granted by the employer, giving employees the right to buy company stock at a specific price during a specific time period, with levels ranging from 0 to 3.

- TotalWorkingYears: The total number of years the employee has been working across all organizations throughout their career.

- TrainingTimesLastYear: The number of training sessions the employee attended in the previous year.

- WorkLifeBalance: Employee satisfaction with work-life balance where 1 means 'Bad', 2 means 'Good', 3 means 'Better', and 4 means 'Best'.

- YearsAtCompany: The number of years the employee has been working in the current organization.

- YearsInCurrentRole: The number of years the employee has been working in their current position.

- YearsSinceLastPromotion: The number of years that have passed since the employee's last promotion.

- YearsWithCurrManager: The number of years the employee has been working under their current manager.

```python

num_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    print(f"Outliers for {col}: {len(outliers)}")
```

## Why We Keep the Outliers in Our HR Analysis

### Decision: Retain All Outliers

After analyzing our dataset, we identified outliers in several key features including **MonthlyIncome** (114 outliers), **PerformanceRating** (226 outliers), **TrainingTimesLastYear** (238 outliers), and **NumCompaniesWorked** (52 outliers). However, we made the strategic decision to **keep all outliers** because they represent meaningful and valuable employee segments rather than data errors.

These outliers actually capture important business scenarios: high earners represent executives and senior roles that are critical to retain, employees with many previous companies indicate job-hopping patterns that strongly predict attrition, and highly trained employees represent significant company investments. Removing these outliers would eliminate exactly the employee segments that HR teams need to understand most for effective retention strategies.

### Technical and Business Justification

From a technical perspective, **Random Forest models are naturally robust to outliers** since they use tree-based splits rather than distance calculations, making our chosen algorithm well-suited to handle these extreme values. More importantly, our analysis showed that these outlier groups exhibit **different attrition patterns** than average employees - for example, job hoppers have significantly higher turnover rates than the company average. By keeping these outliers, our model can learn these distinct patterns and provide more accurate predictions for all employee segments, ultimately delivering better business insights

## Machine Learning: Splitting the Data into Training and Testing Sets

We do not use the entire dataset to build a model. Instead, a portion of the data is randomly set aside to evaluate how well the model performs — this is called the Testing Data. The remaining portion, known as the Training Data, is used to train the model. Typically, around 70% of the data is allocated for training, while the remaining 30% is reserved for testing.

- Your dataset has both numerical (Age, MonthlyIncome) and categorical (Department, JobRole) features

```python
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import GridSearchCV, cross_val_score
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## Prepare Features and Target

```python
# Remove columns that won't help with prediction
columns_to_drop = ['EmployeeNumber', 'EmployeeCount', 'StandardHours', 'Over18']

# Create feature matrix (X) - everything except Attrition and unnecessary columns
X = df.drop(columns_to_drop + ['Attrition'], axis=1)

# Create target vector (y) - this is what we want to predict
y = df['Attrition']

print(f"Features (X) shape: {X.shape}")
print(f"Target (y) shape: {y.shape}")
print(f"\nFeatures included: {list(X.columns)}")
print(f"\nTarget distribution:")
print(y.value_counts())
print(f"Attrition rate: {y.mean():.2%}")
```

## Train-Test Split

```python
# (70% training, 30% testing)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=42, 
    stratify=y  
)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Test set size: {X_test.shape[0]} samples")

print(f"\nTraining set target distribution:")
print(y_train.value_counts())
print(f"Training attrition rate: {y_train.mean():.2%}")

print(f"\nTest set target distribution:")
print(y_test.value_counts())
print(f"Test attrition rate: {y_test.mean():.2%}")
```

## Handle Class Imbalance

```python
# apply oversampling to training data only to balance the classes
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

print(f"Balanced training set size: {X_train_balanced.shape[0]} samples")
print(f"Attrition distribution after SMOTE:\n{pd.Series(y_train_balanced).value_counts()}")
```

 ## Random Forest with Hyperparameter Tuning

```python
print("=== TRAINING RANDOM FOREST MODEL ===")
print("Training in progress... Please wait...")

# define parameter grid for tuning
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [8, 12, 15],
    'min_samples_split': [5, 10],
    'min_samples_leaf': [2, 4, 6],
    'max_features': ['sqrt', 'log2']
}
# initialize
rf_model = RandomForestClassifier(random_state=42, n_jobs=-1, class_weight={0:1, 1:3}
)
# perform Grid Search with Cross-Validation
grid_search = GridSearchCV(
    rf_model,
    param_grid,
    cv=5,  
    scoring='recall',  
    n_jobs=-1,
    verbose=0
)

# train the model
grid_search.fit(X_train_balanced, y_train_balanced)

# get the best model
best_rf_model = grid_search.best_estimator_


print("\nTraining completed!")
print("Best Parameters:")
for k, v in grid_search.best_params_.items():
    print(f"   • {k}: {v}")
print(f"Best CV AUC: {grid_search.best_score_:.4f}")
```

## Make Predictions


```python
# Make predictions on test set
print("=== MAKING PREDICTIONS ===")

y_pred = best_rf_model.predict(X_test)
y_pred_proba = best_rf_model.predict_proba(X_test)[:, 1]

print("Predictions completed!")
print(f"Test set size: {len(y_test)}")
print(f"Predictions made: {len(y_pred)}")
```

## Model Evaluation

```python
y_pred = best_rf_model.predict(X_test)
y_proba = best_rf_model.predict_proba(X_test)[:,1]
cm =confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc_score = roc_auc_score(y_test, y_pred_proba)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"ROC AUC Score: {auc_score:.4f}")
print()
print("CLASSIFICATION REPORT")
print()
print(classification_report(y_test, y_pred))
print("Confusion Matrix:")
print(cm)
```
